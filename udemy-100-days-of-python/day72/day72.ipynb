Sorting Values & Adding Columns

highest starting salary

```python
df.sort_values("Starting Median Salary", ascending=False).head(10)
```

lowest

```python
df.sort_values("Starting Median Salary").head(10)
```

create a new column

```python
df["Salary Difference"] = (
    df["Mid-Career Median Salary"]
    - df["Starting Median Salary"]
)
```

now read

```python
df.head()
```

Degrees with the Highest Potential

```python
df.sort_values("Salary Difference", ascending=False).head(10)
```

calculate the percentage increase

```python
df["Salary Increase %"] = (
    df["Salary Difference"] /
    df["Starting Median Salary"]
) * 100
```

```python
df.sort_values("Salary Increase %", ascending=False).head(10)
```

Grouping and Pivoting

columun representing major category         

```python
df.groupby("Major_category")["Mid-Career Median Salary"].mean()
```

```python
df.groupby("Major_category")["Mid-Career Median Salary"].mean().sort_values(
    ascending=False
)
```

pivot table

```python
pd.pivot_table(
    df,
    values="Mid-Career Median Salary",
    index="Major_category",
    aggfunc="mean"
)
```
