## Mastering Errors, Exceptions, and JSON Data in Python

No application is perfect. Users can enter invalid data, files can go missing, and unexpected errors can occur at any time. On Day 30 of Dr. Angela Yu's 100 Days of Code, I learned how to build more reliable Python applications by handling exceptions gracefully and replacing plain text storage with JSON.

Understanding Exception Handling

```python
try:
    file = open("data.txt")
except FileNotFoundError:
    print("File does not exist.")
else:
    contents = file.read()
    print(contents)
finally:
    print("Program finished.")
```

Raising Your Own Exceptions

```python
height = 1.55

if height < 1.6:
    raise ValueError("Height must be at least 1.6 meters.")
```

Handling a KeyError

```python
data = {
    "France": "Paris",
    "Germany": "Berlin"
}

try:
    print(data["Nigeria"])
except KeyError:
    print("Country not found.")
```

Handling an IndexError

```python
numbers = [1, 2, 3]

try:
    print(numbers[5])
except IndexError:
    print("Index is out of range.")
```

Handling Missing JSON Files

```python
try:
    with open("data.json", "r") as file:
        data = json.load(file)
except FileNotFoundError:
    with open("data.json", "w") as file:
        json.dump(new_data, file, indent=4)
else:
    data.update(new_data)

    with open("data.json", "w") as file:
        json.dump(data, file, indent=4)
```

Searching for a saved password

```python
try:
    with open("data.json", "r") as file:
        data = json.load(file)
except FileNotFoundError:
    print("No data file found.")
else:
    if website in data:
        print(f"Email: {data[website]['email']}")
        print(f"Password: {data[website]['password']}")
    else:
        print(f"No details for {website} exist.")
```
