```python
from flask import Flask, render_template

app = Flask(__name__)

@app.route("/")
def home():
    return render_template("index.html")

if __name__ == "__main__":
    app.run()
```

4. creating index.html page

```python
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">

    <title>My Flask Website</title>
</head>

<body>

    <h1>Welcome to My Website</h1>

    <p>This webpage is being rendered using Flask templates.</p>

</body>
</html>
```

5. passing python variables to HTML

```python
from flask import Flask, render_template

app = Flask(__name__)

@app.route("/")
def home():

    name = "Muhammad"
    age = 25

    return render_template(
        "index.html",
        name=name,
        age=age
    )

if __name__ == "__main__":
    app.run()
```

6. creating dynamic content with ninja

```python
from flask import Flask, render_template

app = Flask(__name__)

@app.route("/")
def home():

    languages = [
        "Python",
        "JavaScript",
        "Java",
        "HTML",
        "CSS"
    ]

    return render_template(
        "index.html",
        languages=languages
    )

if __name__ == "__main__":
    app.run()
```

```python
<!DOCTYPE html>
<html lang="en">

<head>
    <meta charset="UTF-8">
    <title>Programming Languages</title>
</head>

<body>

    <h1>Programming Languages</h1>

    <ul>

        {% for language in languages %}

        <li>{{ language }}</li>

        {% endfor %}

    </ul>

</body>

</html>
```

7. mini dynamic blog website

```python
from flask import Flask, render_template

app = Flask(__name__)

posts = [
    {
        "title": "Learning Python",
        "content": "Python is a powerful and beginner-friendly programming language."
    },

    {
        "title": "Learning Flask",
        "content": "Flask makes it easy to build web applications using Python."
    },

    {
        "title": "100 Days of Python",
        "content": "Consistency is helping me improve my programming skills every day."
    }
]


@app.route("/")
def home():

    return render_template(
        "index.html",
        posts=posts
    )


if __name__ == "__main__":
    app.run()
```
