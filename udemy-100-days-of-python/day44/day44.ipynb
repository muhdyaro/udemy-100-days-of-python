4. CSS Box Model 

```python
div {
    width: 300px;
    height: 200px;

    padding: 20px;
    border: 5px solid black;
    margin: 30px;
}
```

Motivational Poster Website Project

```python
<!DOCTYPE html>
<html lang="en">

<head>
    <meta charset="UTF-8">
    <title>Motivational Poster</title>
    <link rel="stylesheet" href="style.css">
</head>

<body>

    <div class="poster">
        <img
            src="image.jpg"
            alt="Motivational image"
        >

        <h1>KEEP GOING</h1>

        <p>
            Success comes to those who keep moving forward.
        </p>
    </div>

</body>

</html>
```

css

```python
body {
    background-color: black;
    color: white;
    text-align: center;
}

.poster {
    width: 50%;
    margin: 50px auto;
}

.poster img {
    width: 100%;
    border: 5px solid white;
}

h1 {
    font-family: Arial, sans-serif;
    font-size: 40px;
}

p {
    font-family: Georgia, serif;
    font-size: 20px;
}
```
