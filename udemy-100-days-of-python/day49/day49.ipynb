### Automating your Excercise Routine at the Gym

1. Import selenium

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
```

2. Create the Chrome driver

```python
chrome_options = webdriver.ChromeOptions()
chrome_options.add_experimental_option("detach", True)

driver = webdriver.Chrome(options=chrome_options)
```

3. open the gym website

```python
driver.get("website url")
```

4. inspect the website

```python
driver.find_element(By.ID, "name")
```

5. find the exercise input

```python
exercise = driver.find_element(By.ID, "exercise")
```

6 . enter the excercise

```python
exercise.send_keys("Push Ups")
```

7. find the repetation input

```python
reps = driver.find_element(By.ID, "reps")
```

8. enter the number of repetations

```python
reps.send_keys("20")
```

9. find the sets input

```python
sets = driver.find_element(By.ID, "sets")
```

10. enter the number of sets

```python
sets.send_keys("3")
```

11. findn the submit button

```python
submit_button = driver.find_element(By.ID, "submit")
```
