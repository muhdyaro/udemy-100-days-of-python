```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time


chrome_options = webdriver.ChromeOptions()
chrome_options.add_experimental_option("detach", True)

driver = webdriver.Chrome(options=chrome_options)



driver.get("YOUR_DAY_49_WEBSITE_URL")


# -----------------------------
# Workout data
# -----------------------------

workout = {
    "Push Ups": {
        "reps": 40,
        "sets": 3
    },

    "Squats": {
        "reps": 30,
        "sets": 3
    },

    "Pull Ups": {
        "reps": 10,
        "sets": 3
    },

    "Sit Ups": {
        "reps": 25,
        "sets": 3
    }
}


for exercise_name, details in workout.items():

    exercise = driver.find_element(By.ID, "exercise")
    exercise.clear()
    exercise.send_keys(exercise_name)

    reps = driver.find_element(By.ID, "reps")
    reps.clear()
    reps.send_keys(str(details["reps"]))

    sets = driver.find_element(By.ID, "sets")
    sets.clear()
    sets.send_keys(str(details["sets"]))

    submit_button = driver.find_element(By.ID, "submit")
    submit_button.click()

    time.sleep(1)
```
