## Day 1 – Band Name Generator

Explanation

On the first day of the 100 Days of Code Python Bootcamp, I learned the fundamentals of Python programming. The lesson focused on printing output to the console, collecting user input, storing information in variables, and combining strings to create meaningful results.

The project was a simple Band Name Generator. The program asks the user for the city they grew up in and the name of their pet, then combines both answers to generate a creative band name. Although simple, this project introduced the basic building blocks that are used throughout Python programming.

```python
# ------------------------------------------
# Day 1 - Band Name Generator
# 100 Days of Code: Python Bootcamp
# ------------------------------------------

# Display a welcome message to the user.
print("Welcome to the Band Name Generator!")

# Ask the user for the name of the city where they grew up.
city = input("Which city did you grow up in?\n")

# Ask the user for the name of their pet.
pet = input("What is the name of your pet?\n")

# Combine both answers to generate a band name.
band_name = f"{city} {pet}"

# Display the generated band name.
print(f"Your band name could be: {band_name}")
```
