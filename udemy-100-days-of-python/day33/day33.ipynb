### API Endpoints and API Parameteras

Today marked one of the most exciting milestones in the course—I finally stepped into the world of Application Programming Interfaces (APIs). APIs are the backbone of modern software, allowing different applications to communicate and exchange data seamlessly.

An API (Application Programming Interface) acts as a bridge between two software systems. Instead of building everything from scratch, developers can request data or services from another application through predefined endpoints.

Some everyday examples include:

Weather applications
Google Maps
Payment gateways
Social media integrations
AI services like ChatGPT

Understanding APIs opens the door to building applications that interact with real-world data.

API Endpoints and Making Requests

I learned how to make HTTP requests using Python's requests library.

The main concepts included:

Sending GET requests
Understanding API endpoints
Reading server responses
Parsing returned data

Example:

```python
import requests

response = requests.get(url="http://api.open-notify.org/iss-now.json")
response.raise_for_status()

data = response.json()
print(data)
```

#### HTTP Status Codes

server always responds with a status code indicating whether a request succeeded or failed.

Some common codes are:

200 — Success
201 — Resource created
301 — Redirect
400 — Bad Request
401 — Unauthorized
403 — Forbidden
404 — Not Found
500 — Internal Server Error

Rather than assuming every request succeeds, it's important to check the response before using the returned data.

#### Exception Handling

I also learned how to handle API errors more safely.

Instead of writing lots of conditional statements, Python provides:

response.raise_for_status()

This raises an exception automatically if the request wasn't successful, making debugging much easier.

#### Working with JSON

Most APIs return data in JSON (JavaScript Object Notation).

```python
import requests

response = requests.get(url="http://api.open-notify.org/iss-now.json")
response.raise_for_status()

data = response.json()

longitude = data["iss_position"]["longitude"]
latitude = data["iss_position"]["latitude"]

print(longitude)
print(latitude)
```

#### Understanding API Parameters

Another important lesson introduced API parameters.
Instead of receiving the same information every time, parameters allow us to customize requests.

```python
import requests

MY_LAT = 51.507351
MY_LONG = -0.127758

parameters = {
    "lat": MY_LAT,
    "lng": MY_LONG,
    "formatted": 0,
}

response = requests.get(
    "https://api.sunrise-sunset.org/json",
    params=parameters,
)
response.raise_for_status()

data = response.json()

sunrise = data["results"]["sunrise"]
sunset = data["results"]["sunset"]

print(sunrise)
print(sunset)
```
