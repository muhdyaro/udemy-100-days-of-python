## Intermediate HTML

HTML boiler plate

Every HTML page starts with a standard structure.


```python
<!DOCTYPE html>
<html lang="en">

<head>
    <meta charset="UTF-8">
    <title>My Website</title>
</head>

<body>

    <h1>Hello World!</h1>

</body>

</html>
```

the list element

unorderd list

```python
<ul>
    <li>Rice</li>
    <li>Beans</li>
    <li>Pasta</li>
</ul>
```

orderd list

```python
<ol>
    <li>Wake up</li>
    <li>Pray</li>
    <li>Study Python</li>
</ol>
```

nesting and identation

```python
<ul>
    <li>Fruits
        <ul>
            <li>Apple</li>
            <li>Orange</li>
            <li>Mango</li>
        </ul>
    </li>

    <li>Vegetables</li>
</ul>
```

ancor element

ancor tags to create hyper link:

```python
<a href="https://www.google.com">
    Visit Google
</a>
```

open a wesite in a new tab:

```python
<a href="https://www.github.com" target="_blank">
    GitHub
</a>
```

link to another page:

```python
<a href="./about.html">About Me</a>
```
