## Building a Password Manager GUI with Tkinter

One of the biggest lessons in software engineering is solving real-world problems with simple tools. On Day 29 of Dr. Angela Yu's 100 Days of Code, I built a Password Manager desktop application using Python and Tkinter.

This project combines GUI design, file handling, random password generation, dialog boxes, and clipboard functionality into a practical application that helps users securely organize their login credentials.

Creating the Main Window and Canvas

```python
from tkinter import *

window = Tk()
window.title("Password Manager")
window.config(padx=50, pady=50)

canvas = Canvas(width=200, height=200)
logo_img = PhotoImage(file="logo.png")
canvas.create_image(100, 100, image=logo_img)
canvas.grid(row=0, column=1)

window.mainloop()
```

Organizing the Layout with grid()

```python
website_label.grid(row=1, column=0)

website_entry.grid(row=1, column=1, columnspan=2)

email_label.grid(row=2, column=0)
email_entry.grid(row=2, column=1, columnspan=2)

password_label.grid(row=3, column=0)
password_entry.grid(row=3, column=1)

generate_password_button.grid(row=3, column=2)

add_button.grid(row=4, column=1, columnspan=2)
```

Saving Data to a File

```python
def save():
    website = website_entry.get()
    email = email_entry.get()
    password = password_entry.get()

    with open("data.txt", "a") as data_file:
        data_file.write(
            f"{website} | {email} | {password}\n"
        )
```

Displaying a Warning Dialog

```python
from tkinter import messagebox

if len(website) == 0 or len(password) == 0:
    messagebox.showinfo(
        title="Oops",
        message="Please don't leave any fields empty!"
    )
```

Confirming Before Saving

```python
is_ok = messagebox.askokcancel(
    title=website,
    message=f"These are the details entered:\n"
            f"Email: {email}\n"
            f"Password: {password}\n\n"
            "Do you want to save?"
)

if is_ok:
    with open("data.txt", "a") as data_file:
        data_file.write(
            f"{website} | {email} | {password}\n"
        )
```

Generating a Random Password

```python
import random

letters = list("abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ")
numbers = list("0123456789")
symbols = ['!', '#', '$', '%', '&', '(', ')', '*', '+']

password_letters = random.choices(letters, k=8)
password_symbols = random.choices(symbols, k=2)
password_numbers = random.choices(numbers, k=2)

password_list = (
    password_letters +
    password_symbols +
    password_numbers
)

random.shuffle(password_list)

password = "".join(password_list)
```

Displaying the Password and Copying It
import pyperclip

```python
import pyperclip

password_entry.delete(0, END)
password_entry.insert(0, password)

pyperclip.copy(password)
```

Clearing the Input Fields After Saving

```python
website_entry.delete(0, END)
password_entry.delete(0, END)
```
