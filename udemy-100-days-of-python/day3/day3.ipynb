Lesson – Control Flow with if, else, and Conditional Operators

Conditional statements allow a program to execute different blocks of code depending on whether a condition is True or False.

```python
# Ask the user for their age.
age = int(input("Enter your age: "))

# Check if the user is old enough to vote.
if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote yet.")
```

Lesson 24 – Nested if Statements and elif

Sometimes one condition depends on another.

Python allows one if statement inside another.

```python
height = int(input("Enter your height in cm: "))

if height >= 120:
    age = int(input("Enter your age: "))

    if age < 12:
        print("Child Ticket: $5")
    elif age <= 18:
        print("Youth Ticket: $7")
    else:
        print("Adult Ticket: $12")

else:
    print("Sorry, you are too short to ride.")
```

Coding Exercise – BMI Calculator with Interpretations

```python
# Ask the user for their height.
height = float(input("Enter your height in meters: "))

# Ask the user for their weight.
weight = float(input("Enter your weight in kilograms: "))

# Calculate BMI.
bmi = weight / (height ** 2)

print(f"Your BMI is {bmi:.1f}")

# Interpret the BMI result.
if bmi < 18.5:
    print("You are underweight.")

elif bmi < 25:
    print("You have a normal weight.")

elif bmi < 30:
    print("You are slightly overweight.")

elif bmi < 35:
    print("You are obese.")

else:
    print("You are clinically obese.")
```

Lesson  – Multiple if Statements in Succession

```python
bill = 0

height = int(input("Enter your height in cm: "))

if height >= 120:

    age = int(input("Enter your age: "))

    if age < 12:
        bill = 5

    elif age <= 18:
        bill = 7

    else:
        bill = 12

    # Ask whether the user wants a photo.
    wants_photo = input("Do you want a photo? Y or N: ")

    if wants_photo == "Y":
        bill += 3

    print(f"Your final bill is ${bill}")

else:
    print("Sorry, you cannot ride.")
```

Lesson 26 – Pizza Order Practice
This project combines everything learned today.

The final bill depends on:

Pizza size
Pepperoni
Extra cheese

```python
print("Welcome to Python Pizza Deliveries!")

bill = 0

size = input("Choose a pizza size (S, M, L): ")

if size == "S":
    bill = 15

elif size == "M":
    bill = 20

elif size == "L":
    bill = 25

pepperoni = input("Add pepperoni? (Y/N): ")

if pepperoni == "Y":

    if size == "S":
        bill += 2

    else:
        bill += 3

extra_cheese = input("Add extra cheese? (Y/N): ")

if extra_cheese == "Y":
    bill += 1

print(f"Your final bill is ${bill}")
```
