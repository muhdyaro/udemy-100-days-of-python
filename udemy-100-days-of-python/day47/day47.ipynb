```python
def send_email(product_title, price):

    message = (
        f"Subject: Amazon Price Alert!\n\n"
        f"{product_title}\n"
        f"is now ${price:.2f}.\n\n"
        f"Buy it here:\n{URL}"
    )

    with smtplib.SMTP("smtp.gmail.com", 587) as connection:

        connection.starttls()

        connection.login(
            EMAIL_ADDRESS,
            EMAIL_PASSWORD
        )

        connection.sendmail(
            EMAIL_ADDRESS,
            EMAIL_ADDRESS,
            message
        )
```
