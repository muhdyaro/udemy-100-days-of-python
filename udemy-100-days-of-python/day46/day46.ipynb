## Creating a Spotify Playlist using the Musical Time Machine

the goal for today is to build a project that combines web scraping, APIs, authentication, and automation to create a music playlist automatically.

Scraping the billboard Hot 100

1. setting up vs code

2. ceate a virtual enviroment

3 getting the billboard date:

```python
import requests
from bs4 import BeautifulSoup

date = input(
    "Which year do you want to travel to? Type the date in this format YYYY-MM-DD: "
)

billboard_url = f"https://www.billboard.com/charts/hot-100/{date}/"

response = requests.get(billboard_url)

print(response.status_code)
```

4, parse the billboard page

```python
soup = BeautifulSoup(response.text, "html.parser")
```

find the song title:

```python
song_elements = soup.select("h3.c-title")

song_names = [
    song.get_text(strip=True)
    for song in song_elements
]

print(f"Found {len(song_names)} songs.")

for song in song_names:
    print(song)
```

5. create the youtube musci api client

Import ytmusicapi:

```python
from ytmusicapi import YTMusic
```

run "ytmusicapi browser" in bash

```python
from ytmusicapi import YTMusic

ytmusic = YTMusic("headers_auth.json")
```

6. create the playlist

```python
playlist_title = f"{date} Billboard 100"

playlist_description = (
    f"Top 100 songs from the Billboard Hot 100 on {date}."
)
```

```python
playlist_id = ytmusic.create_playlist(
    playlist_title,
    playlist_description
)

print("Playlist created!")
```

7. search for the songs

```python
song_ids = []

for song in song_names:
    search_results = ytmusic.search(
        song,
        filter="songs"
    )

    if search_results:
        video_id = search_results[0]["videoId"]
        song_ids.append(video_id)

        print(f"Found: {song}")
```

8. add songs to the playlist

```python
ytmusic.add_playlist_items(
    playlist_id,
    song_ids
)
```
