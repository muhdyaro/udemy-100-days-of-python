Pivoting DataFrames

we chanje to a new readable data frame

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("QueryResults.csv")
```

```python
df["m"] = pd.to_datetime(df["m"])
```

```python
print(df.columns)
```

```python
print(df.head())
```

```python
pivoted_df = df.pivot(
    index="m",
    columns="TagName",
    values="Count"
)
```

```python
pivoted_df.head()
```

the new data frames

```python
pivoted_df.shape
```

```python
pivoted_df.columns
```

Data Visualisation with Matplotlib

```python
pip install matplotlib
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(
    pivoted_df.index,
    pivoted_df["python"]
)

plt.show()
```

add another label

```python
plt.figure(figsize=(12, 6))

plt.plot(
    pivoted_df.index,
    pivoted_df["python"]
)

plt.xlabel("Year")
plt.ylabel("Number of Questions")
plt.title("Popularity of Python on Stack Overflow")

plt.show()
```

Multi-Line Charts

we can compare multiple programming languages

```python
plt.figure(figsize=(12, 6))

plt.plot(
    pivoted_df.index,
    pivoted_df["python"],
    label="Python"
)

plt.plot(
    pivoted_df.index,
    pivoted_df["java"],
    label="Java"
)

plt.plot(
    pivoted_df.index,
    pivoted_df["javascript"],
    label="JavaScript"
)

plt.legend()

plt.xlabel("Year")
plt.ylabel("Number of Questions")
plt.title("Programming Language Popularity")

plt.show()
```
