creatiing a beautiful soup object:

```python
html = """
<html>
    <body>
        <h1>Hello World</h1>
        <p>This is my website.</p>
    </body>
</html>
"""
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html, "html.parser")
```

#### Finding and Selecting Particular Elements with BeautifulSoup

1. the .find() method finds the first matching element.

```python
<h1>Welcome to My Website</h1>
<h1>About Me</h1>
<h1>Contact</h1>
```

python:

```python
heading = soup.find("h1")
print(heading)
```

to get the text:

```python
print(heading.getText())
```

2. If you want all matching elements, use .find_all().

```python
headings = soup.find_all("h1")
```

looping throgh item:

```python
for heading in headings:
    print(heading.getText())
```

3. Finding elemets class

suppose html is:

<h2 class="movie-title">The Dark Knight</h2>
<h2 class="movie-title">Inception</h2>
<h2 class="movie-title">Interstellar</h2>

we can search for class:

```python
movies = soup.find_all(class_="movie-title")
```
