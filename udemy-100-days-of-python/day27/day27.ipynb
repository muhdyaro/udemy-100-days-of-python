## TKinter, *args, **kwargs and Creating GUI Programs

Introduction to Tkinter
Tkinter is Python's standard library for building desktop applications. It provides everything needed to create graphical interfaces, including windows, buttons, labels, text fields, checkboxes, and more.
Creating a basic application window is straightforward.

```python
from tkinter import * 
window = Tk() 
window.title("My First GUI") 
window.minsize(width=500, height=300) 
window.mainloop()
```

Creating Labels

Labels display text inside the application.

```python
from tkinter import * 
window = Tk() 
label = Label(text="Hello, World!", font=("Arial", 24, "bold")) 
label.pack() 
window.mainloop()
```

Default Function Arguments

Python functions can define default parameter values, making arguments optional.

```python
def greet(name="Guest"): 
    print(f"Hello, {name}") 
greet() 
greet("Muhammad")
```

Using *args

Sometimes you don't know how many positional arguments a function will receive.

That's where *args becomes useful.

```python
def add(*args): 
    return sum(args) 
print(add(2, 4, 6)) 
print(add(1, 2, 3, 4, 5))
```

Using **kwargs

While *args collects positional arguments, **kwargs collects keyword arguments into a dictionary.

```python
def student(**kwargs): 
    for key, value in kwargs.items(): 
        print(key, value) 
student(name="Muhammad", age=24, country="Nigeria")


# Because kwargs behaves like a dictionary, values can be accessed by their keys
def student(**kwargs): 
    print(kwargs["name"])
```

Buttons and Entry Widgets

Tkinter applications become interactive by combining widgets with functions.

Creating a button:

```python
from tkinter import * 
window = Tk() 

def button_clicked(): 
    print("Button clicked!") 

button = Button(text="Click Me", command=button_clicked) 
button.pack() 
window.mainloop()
```

Tkinter Layout Managers

Tkinter offers three different ways to position widgets.

pack()
label.pack()

A simple layout manager that stacks widgets automatically.

place()
label.place(x=100, y=80)

Allows precise positioning using x and y coordinates.
