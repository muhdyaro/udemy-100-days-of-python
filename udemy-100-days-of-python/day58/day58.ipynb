## Web Foundation Bootstrap

1. what is boostrap

ans: bootstrap is a front-end framework containing ready-made CSS classes and components.

instead of writing all your CSS from scratch, you can use Bootstrap classes such as:

```python
<button class="btn btn-primary">Click Me</button>
```

2. setting up boostrap

add the bootstrap CDN to HTML <head> section:

```python
<link href="https://cdn.jsdelivr.net/npm/bootstrap@5.3.3/dist/css/bootstrap.min.css" rel="stylesheet">  
```

add the javaScript bundle before the closing </body> tag:

```python
<script src="https://cdn.jsdelivr.net/npm/bootstrap@5.3.3/dist/js/bootstrap.bundle.min.js"></script>
```

3. bootstrap containers

```python
<div class="container">

    <h1>Welcome!</h1>

    <p>This content is inside a Bootstrap container.</p>

</div>
```
