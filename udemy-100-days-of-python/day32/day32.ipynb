### Send Email (SMTP) & Manage Dates

How to Send Emails with Python using SMTP
What is SMTP?

SMTP stands for Simple Mail Transfer Protocol.

It is the protocol email providers use to send emails.

Python communicates with an SMTP server to send messages.

Example:

Python
      ↓
SMTP Server
      ↓
Recipient

```python
# Python Module
import smtplib
```

```python
# Create a connection
connection = smtplib.SMTP("smtp.gmail.com")
```

```python
#Start encryption
connection.starttls()   
```

```python

# Login
connection.login(
    user="your_email",
    password="app_password"
)

# Send mail
connection.sendmail(
    from_addr="you@gmail.com",
    to_addrs="friend@gmail.com",
    msg="Subject:Hello\n\nThis is my first Python email!"
)

# Close connection
connection.close()

# Or use
with smtplib.SMTP("smtp.gmail.com") as connection:
    ...
```

#### Working with datetime

```python
#Python include
from datetime import datetime

#Current date
now = datetime.now()
print(now)
```

```python
# Currrent year
print(now.year)
```

```python
# Current month
print(now.month)
```

```python
# Current weekday
print(now.weekday())
```

Creating your own date

```python
birthday = datetime(
    year=2002,
    month=5,
    day=10
)
print(birthday)
```
