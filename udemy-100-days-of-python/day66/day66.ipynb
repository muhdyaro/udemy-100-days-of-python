using request to communicate with API

```python
import requests

response = requests.get("http://127.0.0.1:5000/all")

print(response.status_code)
print(response.json())
```

search for a cafe

```python
response = requests.get(
    "http://127.0.0.1:5000/search",
    params={"id": 1}
)

print(response.status_code)
print(response.json())
```

posst a new cafe

```python
new_cafe = {
    "name": "Arewa Coffee",
    "map_url": "https://maps.google.com",
    "img_url": "https://example.com/cafe.jpg",
    "location": "Kaduna",
    "seats": "20-30",
    "has_toilet": True,
    "has_wifi": True,
    "has_sockets": True,
    "can_take_calls": False,
    "coffee_price": "$2.50"
}

response = requests.post(
    "http://127.0.0.1:5000/add",
    json=new_cafe
)

print(response.status_code)
print(response.json())
```

path

```python
update_data = {
    "coffee_price": "$3",
    "has_wifi": True
}

response = requests.patch(
    "http://127.0.0.1:5000/update/1",
    json=update_data
)

print(response.status_code)
print(response.json())
```

delete

```python
response = requests.delete(
    "http://127.0.0.1:5000/delete/1"
)

print(response.status_code)
print(response.json())
```
