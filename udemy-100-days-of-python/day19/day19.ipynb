Getting Started with Turtle Graphics

Python's built-in turtle module allows us to create drawings using a virtual turtle that moves around the screen.

```python
# Import the Turtle class and Screen class.
from turtle import Turtle, Screen

# Create a turtle object.
tim = Turtle()

# Move the turtle forward.
tim.forward(100)

# Keep the window open.
screen = Screen()
screen.exitonclick()
```

Drawing Different Shapes

By combining loops and turning angles, the turtle can draw polygons.

```python
from turtle import Turtle, Screen

tim = Turtle()

# Draw shapes from triangle to decagon.
for sides in range(3, 11):

    angle = 360 / sides

    for _ in range(sides):
        tim.forward(100)
        tim.right(angle)

screen = Screen()
screen.exitonclick()
```

Random Walk

The turtle moves in random directions, producing an unpredictable path.

```python
from turtle import Turtle, Screen
import random

tim = Turtle()

directions = [0, 90, 180, 270]

tim.speed("fastest")
tim.pensize(8)

for _ in range(200):

    tim.forward(30)

    tim.setheading(random.choice(directions))

screen = Screen()
screen.exitonclick()
```

Generating Random Colors

```python
import random
from turtle import colormode

# Enable RGB colors.
colormode(255)

# Function to generate a random color.
def random_color():

    r = random.randint(0,255)
    g = random.randint(0,255)
    b = random.randint(0,255)

    return (r,g,b)  
```

Drawing a Spirograph

Combining loops with random colors creates beautiful geometric art.

```python
from turtle import Turtle, Screen, colormode
import random

tim = Turtle()

colormode(255)

tim.speed("fastest")

def random_color():

    return (
        random.randint(0,255),
        random.randint(0,255),
        random.randint(0,255)
    )

for angle in range(0,360,10):

    tim.color(random_color())

    tim.circle(100)

    tim.setheading(angle)

screen = Screen()
screen.exitonclick()
```

Event Listeners

Programs can respond to keyboard input using event listeners.

```python
from turtle import Turtle, Screen

tim = Turtle()

screen = Screen()

# Move forward when the Up arrow is pressed.
def move_forward():
    tim.forward(20)

screen.listen()

screen.onkey(move_forward, "Up")

screen.exitonclick()
```

Project – Etch-A-Sketch App

```python
from turtle import Turtle, Screen

tim = Turtle()

screen = Screen()

# Move forward.
def forward():
    tim.forward(20)

# Move backward.
def backward():
    tim.backward(20)

# Turn left.
def turn_left():
    tim.left(10)

# Turn right.
def turn_right():
    tim.right(10)

# Clear the drawing.
def clear():
    tim.clear()
    tim.penup()
    tim.home()
    tim.pendown()

screen.listen()

screen.onkey(forward, "w")
screen.onkey(backward, "s")
screen.onkey(turn_left, "a")
screen.onkey(turn_right, "d")
screen.onkey(clear, "c")

screen.exitonclick()
```
