## Understanding the Tkinter GUI, Creating Windows, and Building a Mile to Kilometer Converter

Introduction to Tkinter

Tkinter is Python's standard library for creating desktop graphical user interfaces (GUIs).

```python
# Import the Tkinter library.
from tkinter import *

# Create the main application window.
window = Tk()

# Set the title of the window.
window.title("My First GUI")

# Set the window size.
window.minsize(width=500, height=300)

# Keep the window open.
window.mainloop()
```

Creating Labels

A Label displays text inside the application window.

```python
from tkinter import *

window = Tk()
window.title("Labels Example")

# Create a label widget.
label = Label(text="Hello, Tkinter!", font=("Arial", 18))

# Display the label.
label.pack()

window.mainloop()
```

Buttons and Event Handling

Buttons perform actions when clicked.

```python
from tkinter import *

window = Tk()

# Function executed when the button is clicked.
def button_clicked():
    print("Button was clicked!")

# Create the button.
button = Button(text="Click Me", command=button_clicked)

button.pack()

window.mainloop()
```

Entry Widget

The Entry widget allows users to type text into the application.

```python
from tkinter import *

window = Tk()

entry = Entry(width=25)

entry.pack()

window.mainloop()
```

Layout Managers

Tkinter provides different ways to position widgets.

```python
from tkinter import *

window = Tk()

label1 = Label(text="Miles")
label2 = Label(text="Km")

label1.grid(column=2, row=0)
label2.grid(column=2, row=1)

window.mainloop()
```

Project – Mile to Kilometer Converter

The final project combines labels, entry fields, buttons, and event handling to create a functional desktop application.

```python
from tkinter import *

# ----------------------------
# Create the application window.
# ----------------------------
window = Tk()
window.title("Mile to Kilometer Converter")
window.config(padx=20, pady=20)

# ----------------------------
# Function to perform conversion.
# ----------------------------
def convert():

    miles = float(miles_input.get())

    kilometers = miles * 1.609

    result_label.config(text=f"{kilometers:.2f}")

# ----------------------------
# Entry field.
# ----------------------------
miles_input = Entry(width=10)

miles_input.grid(column=1, row=0)

# ----------------------------
# Labels.
# ----------------------------
miles_label = Label(text="Miles")

miles_label.grid(column=2, row=0)

equal_label = Label(text="is equal to")

equal_label.grid(column=0, row=1)

result_label = Label(text="0")

result_label.grid(column=1, row=1)

km_label = Label(text="Km")

km_label.grid(column=2, row=1)

# ----------------------------
# Button.
# ----------------------------
calculate_button = Button(
    text="Calculate",
    command=convert
)

calculate_button.grid(column=1, row=2)

# ----------------------------
# Run the application.
# ----------------------------
window.mainloop()
```
