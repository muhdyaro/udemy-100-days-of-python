2. import selenium

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
```

3. start  chrome

```python
chrome_options = webdriver.ChromeOptions()
chrome_options.add_argument("--start-maximized")

driver = webdriver.Chrome(options=chrome_options)
```

4. open practice form

```python
driver.get("https://www.selenium.dev/selenium/web/web-form.html")

time.sleep(3)
```

5. fill the text field

```python
name_field = driver.find_element(By.NAME, "my-text")
name_field.send_keys("Muhammad Yaro")
```

6. fill the password field

```python
password_field = driver.find_element(By.NAME, "my-password")
password_field.send_keys("Python123")
```

7. select the dropdown

```python
from selenium.webdriver.support.ui import Select

dropdown = Select(driver.find_element(By.NAME, "my-select"))

dropdown.select_by_visible_text("One")
```

8. submit the form

```python
submit_button = driver.find_element(By.CSS_SELECTOR, "button")
submit_button.click()

time.sleep(3)
```

9. check the result

```python
print("Current URL:", driver.current_url)driver.quit()
print("Page title:", driver.title)
```

10. close browser

```python
driver.quit()
```
