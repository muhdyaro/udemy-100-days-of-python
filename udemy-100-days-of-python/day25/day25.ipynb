he Great Squirrel Census Data Analysis

Today's first project involved analyzing real-world data collected from the Central Park Squirrel Census.

The objective was to count squirrels based on their primary fur color.

```python
import pandas as pd

# Load the squirrel dataset.
data = pd.read_csv("weather.csv")

# Count squirrels by fur color.
gray = len(data[data["Primary Fur Color"] == "Gray"])

black = len(data[data["Primary Fur Color"] == "Black"])

cinnamon = len(data[data["Primary Fur Color"] == "Cinnamon"])

# Create a summary table.
summary = {
    "Fur Color": ["Gray", "Black", "Cinnamon"],
    "Count": [gray, black, cinnamon]
}

result = pd.DataFrame(summary)

print(result)

# Save the summary to a new CSV file.
result.to_csv("squirrel_count.csv")
```

U.S. States Game (Part 1)

The second project combines Pandas with Turtle Graphics.

The program displays a blank map of the United States and prompts the player to guess state names.

```python
from turtle import Screen
import pandas as pd

# Create the game window.
screen = Screen()

screen.title("U.S. States Game")

# Load the state dataset.
data = pd.read_csv("50_states.csv")
```

Checking User Answers

Each time the user enters a state name, the program checks whether the answer exists in the dataset.


```python
answer = screen.textinput(
    title="Guess the State",
    prompt="Enter a state name:"
)

if answer in data.state.values:
    print("Correct!")
```

Finding Coordinates

Each state has x and y coordinates stored in the CSV file.

```python
state_data = data[data.state == answer]

x = int(state_data.x)

y = int(state_data.y)

print(x, y)
```

Saving Missing States

When the user exits the game, every missed state is saved into a new CSV file for future practice.

```python
# Convert all states into a list.
all_states = data.state.to_list()

# Store guessed states.
guessed_states = ["Texas", "California", "Florida"]

# Determine missing states.
missing_states = []

for state in all_states:

    if state not in guessed_states:
        missing_states.append(state)

# Convert the list into a DataFrame.
missing = pd.DataFrame(missing_states)

# Save the file.
missing.to_csv("states_to_learn.csv")
```
