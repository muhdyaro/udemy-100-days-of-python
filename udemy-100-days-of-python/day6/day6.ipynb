Functions
A function is a named block of code that performs a specific task.

Instead of repeating the same code, we define a function once and call it whenever we need it.

Example

```python
# Define a function.
def greet():
    print("Hello!")
    print("Welcome to Python.")
    print("Happy Coding!")

# Call the function.
greet()
```

## Creating a Custom Function

The robot only knows how to turn left.
To make it turn right, I created my own function.


💻 Example

```python
# Create a custom function.

def turn_right():
    turn_left()
    turn_left()
    turn_left()
```

Now, instead of writing three turn_left() commands every time, I can simply write:

```python
turn_right()
```

While Loops

A while loop repeats a block of code as long as a condition remains True.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

Project – Escaping the Maze

The final project required programming the robot to solve different maze layouts.

The strategy was:

Move forward whenever possible.
Turn right if the path is blocked.
Turn left when necessary.
Continue until reaching the goal.

```python
# Create a helper function to turn right.

def turn_right():
    turn_left()
    turn_left()
    turn_left()

# Continue moving until the robot reaches the goal.
while not at_goal():

    # If there is space ahead, keep moving.
    if front_is_clear():
        move()

    # Otherwise, turn right.
    else:
        turn_right()

    # If the right side is clear after turning,
    # continue moving.
    while right_is_clear():
        turn_right()
        move()
```

Another Maze Strategy

Some maze layouts require following the wall.

```python
def turn_right():
    turn_left()
    turn_left()
    turn_left()

while not at_goal():

    if right_is_clear():
        turn_right()
        move()

    elif front_is_clear():
        move()

    else:
        turn_left()
```
