# Building a Turtle Game

Today, I built a Turtle Crossing Game using Python’s Turtle module. This project brought together several concepts I’ve been learning, especially Object-Oriented Programming (OOP), event handling, collision detection, and game logic.

One of the first things I implemented was player movement using keyboard input:

```python
screen.listen() screen.onkey(player.move, "Up")
```

This allowed the turtle to move forward whenever the Up arrow key was pressed.

To create traffic, I used a car manager that continuously generates cars at random positions:

```python
new_car = Turtle("square") new_car.penup() new_car.goto(300, random.randint(-250, 250))
```

I also learned how collision detection works by checking the distance between the player and each car:

```python
for car in car_manager.all_cars: if car.distance(player) < 20: game_is_on = False
```

This simple logic made the game end whenever the turtle was hit by a car.

Another interesting feature was increasing the difficulty as the player advanced through levels:

```python
if player.is_at_finish_line(): player.go_to_start() car_manager.level_up() scoreboard.increase_level()
```

This made the cars move faster each time the player successfully crossed the road.
