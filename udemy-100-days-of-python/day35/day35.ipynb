#### Check if It Will Rain in the Next 12 Hours

In the previous lesson i successfully connected the openwhethermap API, authenticate the API key and retrieve whether forecast data. in this lesson i build core logic of the Rain alert project by checking the whether forecast for the next 12 hours and determing whether you should carry an umbrella.

```python
import requests

API_KEY = "3d963377a57efe98f0e91427c6ea1b7a"

OWM_ENDPOINT = "https://api.openweathermap.org/data/2.5/forecast"

MY_LAT = -51.5774
MY_LONG = -19.1046

weather_params = {
    "lat": MY_LAT,
    "lon": MY_LONG,
    "appid": API_KEY,
    "cnt": 4,
}

response = requests.get(
    url=OWM_ENDPOINT,
    params=weather_params
)

response.raise_for_status()

weather_data = response.json()

will_rain = False

for hour_data in weather_data["list"]:
    condition_code = hour_data["weather"][0]["id"]

    print(f"Weather ID: {condition_code}")

    if condition_code < 700:
        will_rain = True

if will_rain:
    print("Bring an Umbrella ☔")
else:
    print("No rain expected today.")
```

```python
import requests
from twilio.rest import Client

API_KEY = ""

ACCOUNT_SID = ""

AUTH_TOKEN = ""

MY_LAT = 10.5105

MY_LONG = 7.4165

OWM_ENDPOINT = "https://api.openweathermap.org/data/2.5/forecast"

weather_params = {
    "lat": MY_LAT,
    "lon": MY_LONG,
    "appid": API_KEY,
    "cnt": 4,
}

response = requests.get(
    url=OWM_ENDPOINT,
    params=weather_params
)

response.raise_for_status()

weather_data = response.json()

will_rain = False

for hour_data in weather_data["list"]:

    condition_code = hour_data["weather"][0]["id"]

    if condition_code < 700:
        will_rain = True

if will_rain:

    client = Client(
        ACCOUNT_SID,
        AUTH_TOKEN
    )

    message = client.messages.create(
        body="☔ It is going to rain today.\nRemember to bring an umbrella!",
        from_="+15005550006",
        to="+2348012345678"
    )

    print(message.status)
else:
    print("No rain expected today.")
```

#### Can't Use SMS? Try WhatsApp Instead

In the previous lesson, i learned how to send an SMS notification using the Twilio API whenever rain is expected. However, in some countries or on Twilio trial accounts, sending SMS messages may not be available or may incur costs.

Twilio also provides a WhatsApp Sandbox, allowing  to send WhatsApp messages through its API. In this lesson, i modify the Rain Alert project to send a WhatsApp notification instead of an SMS.

```python
# Instaall the Stilio SDK
pip install swilio
```

```python
import requests
from twilio.rest import Client

API_KEY = "YOUR_OPENWEATHER_API_KEY"

ACCOUNT_SID = ""

AUTH_TOKEN = ""

OWM_ENDPOINT = "https://api.openweathermap.org/data/2.5/forecast"

MY_LAT = 10.5105
MY_LONG = 7.4165

weather_params = {
    "lat": MY_LAT,
    "lon": MY_LONG,
    "appid": API_KEY,
    "cnt": 4,
}

response = requests.get(
    url=OWM_ENDPOINT,
    params=weather_params
)

response.raise_for_status()

weather_data = response.json()

will_rain = False

for hour_data in weather_data["list"]:

    condition_code = hour_data["weather"][0]["id"]

    if condition_code < 700:
        will_rain = True

if will_rain:

    client = Client(
        ACCOUNT_SID,
        AUTH_TOKEN
    )

    message = client.messages.create(

        body="☔ Rain Alert!\nBring an umbrella today.",

        from_="whatsapp:",

        to="whatsapp:"

    )

    print(message.status)

else:

    print("No rain expected today.")
```
