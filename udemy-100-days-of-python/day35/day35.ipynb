On this day i learred what is API Authentication, how API wroks, how to send authentication API request and why we need to authenticate ourselves 

```python
import requests

API_KEY = "32i43894"
MY_LAT = 10.5105
MY_LONG = 7.4165

parameters = {
    "lat": MY_LAT,
    "lon": MY_LONG,
    "appid": API_KEY,
    "cnt": 4,
}

response = requests.get(
    "https://api.openweathermap.org/data/2.5/forecast",
    params=parameters
)
response.raise_for_status()

weather_data = response.json()

will_rain = False

for hour_data in weather_data["list"]:
    condition_code = hour_data["weather"][0]["id"]

    if int(condition_code) < 700:
        will_rain = True

if will_rain:
    print("Bring an umbrella ☔")
else:
    print("No rain expected.")
```

#### Check if It Will Rain in the Next 12 Hours

In the previous lesson i successfully connected the openwhethermap API, authenticate the API key and retrieve whether forecast data. in this lesson i build core logic of the Rain alert project by checking the whether forecast for the next 12 hours and determing whether you should carry an umbrella.

```python
import requests

API_KEY = ""

OWM_ENDPOINT = "https://api.openweathermap.org/data/2.5/forecast"

MY_LAT = 10.5105
MY_LONG = 7.4165

weather_params = {
    "lat": MY_LAT,
    "lon": MY_LONG,
    "appid": API_KEY,
    "cnt": 4,
}

response = requests.get(
    url=OWM_ENDPOINT,
    params=weather_params
)

response.raise_for_status()

weather_data = response.json()

will_rain = False

for hour_data in weather_data["list"]:
    condition_code = hour_data["weather"][0]["id"]

    print(f"Weather ID: {condition_code}")

    if condition_code < 700:
        will_rain = True

if will_rain:
    print("Bring an Umbrella ☔")
else:
    print("No rain expected today.")
```

Instead of printing only the weather ID, print the description.

```python
for hour_data in weather_data["list"]:
    description = hour_data["weather"][0]["description"]
    print(description)
```

The API also returns a forecast timestamp.
