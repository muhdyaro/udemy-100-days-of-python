### Add a Pixel to the Habit Tracker Using a POST Request

##### After creating our Pixela account and graph, the next step is to start tracking our daily progress.
to create a pixel, we'll send another HTTP POST request. This time, instead of creating a user or a graph, we're creating a daily data entry that contains the date and the quantity of the habit completed.

```python
#import libraries
import requests 
from datetime import datetime 

#define credentiald
USERNAME = "muhdayaro" 
TOKEN = "muhd0903402286" 
GRAPH_ID = "graph1" 

#create the pixel endpoint
PIXEL_ENDPOINT = (
     f"https://pixe.la/v1/users/{USERNAME}/graphs/{GRAPH_ID}"
       ) 

#create the authentication header
headers = { 
    "X-USER-TOKEN": TOKEN 
    } 

#get todays date
today = datetime.now() 

#create the poxel data
pixel_data = { 
    "date": today.strftime("%Y%m%d"), "quantity": "5" 
    } 

#send post request
response = requests.post( 
    url=PIXEL_ENDPOINT, json=pixel_data, headers=headers
      ) 

#display response
print(response.text)
```

in this lessson improve the project by generating todays date with pythonsstrftime() method, making the habit tracker even more convenient to use

### Autofilling Today's Date Using strftime()

using strftime(), our program can automatically generate today's date in the format required by the pixela api (YYYYMMDD). this means we can run the script every day without editing the code, making it perfect for automation with tools like github actions or scheduled tasks.

```python
import requests
from datetime import datetime

USERNAME = "muhdayaro"
TOKEN = "muhd0903402286"
GRAPH_ID = "graph1"

PIXEL_ENDPOINT = (
    f"https://pixe.la/v1/users/{USERNAME}/graphs/{GRAPH_ID}"
)

headers = {
    "X-USER-TOKEN": TOKEN
}

pixel_data = {
    "date": datetime.now().strftime("%Y%m%d"),
    "quantity": "5"
}

response = requests.post(
    url=PIXEL_ENDPOINT,
    json=pixel_data,
    headers=headers
)

print(response.text)
```

### How to Use HTTP PUT and DELETE Requests

in the previous lessons, we learned how to create a Pixela account, create a graph, and add daily habit records using HTTP POST requests. however, real-world applications often require modifying or removing existing data. this is where the PUT and DELETE HTTP methods become essential.

theres is CRUD operations which are used as folllows

theres is CRUD operations which are used as folllows
TTTP method           CRUD operation                   purpose              
GET                    read                         retrieve data from the server
POST                   create                       add new date
PUT                    update                       modify existing data 
DELETE                 delete                       remove existing data

```python
import requests 
from datetime import datetime 

USERNAME = "muhdayaro" 
TOKEN = "muhd0903402286" 
GRAPH_ID = "graph1" 

headers = { 
    "X-USER-TOKEN": TOKEN 
    } 

date = datetime.now().strftime("%Y%m%d") 

PIXEL_ENDPOINT = ( 
    f"https://pixe.la/v1/users/{USERNAME}/graphs/{GRAPH_ID}/{date}" 
    ) 

new_pixel_data = { 
    "quantity": "7" 
    } 

response = requests.put( 
    url=PIXEL_ENDPOINT, json=new_pixel_data, headers=headers 
    ) 

print(response.text)
```
