#### HTTP POST Requests

In the previous lessons, we primarily used HTTP GET requests to retrieve information from APIs such as OpenWeatherMap, Alpha Vantage, and the Open Trivia Database. However, REST APIs support several HTTP methods, each designed for a specific purpose. In this lesson, we'll focus on the POST request, which is used to send new data to a server.    

```python
import requests 
PIXELA_ENDPOINT = "https://pixe.la/v1/users" 
user_params = { 
    "token": "", "username": "muhdayaro", "agreeTermsOfService": "yes", "notMinor": "yes" 
    } 
response = requests.post(
     url=PIXELA_ENDPOINT, json=user_params 
     ) 
print(response.text)
```

Advanced Authentication Using HTTP Headers

```python
import requests 
USERNAME = "" 
TOKEN = "" 

GRAPH_ENDPOINT = f"https://pixe.la/v1/users/{USERNAME}/graphs" 
headers = {
     "X-USER-TOKEN": TOKEN 
     } 

graph_config = {
     "id": "graph1", 
     "name": "Coding Graph", 
     "unit": "hours", 
     "type": "float", 
     "color": "shibafu" 
     } 
response = requests.post( 
    url=GRAPH_ENDPOINT, 
    json=graph_config, 
    headers=headers ) 
print(response.text)
```

### Add a Pixel to the Habit Tracker Using a POST Request

##### After creating our Pixela account and graph, the next step is to start tracking our daily progress.
To create a pixel, we'll send another HTTP POST request. This time, instead of creating a user or a graph, we're creating a daily data entry that contains the date and the quantity of the habit completed.

```python
#import libraries
import requests 
from datetime import datetime 

#define credentiald
USERNAME = "" 
TOKEN = "" 
GRAPH_ID = "graph1" 

#create the pixel endpoint
PIXEL_ENDPOINT = (
     f"https://pixe.la/v1/users/{USERNAME}/graphs/{GRAPH_ID}"
       ) 

#create the authentication header
headers = { 
    "X-USER-TOKEN": TOKEN 
    } 

#get todays date
today = datetime.now() 

#create the poxel data
pixel_data = { 
    "date": today.strftime("%Y%m%d"), "quantity": "5" 
    } 

#send post request
response = requests.post( 
    url=PIXEL_ENDPOINT, json=pixel_data, headers=headers
      ) 

#display response
print(response.text)
```
