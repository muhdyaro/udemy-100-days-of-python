### Building a Habit Tracker with HTTP Requests and the Pixela API

I explored the different types of HTTP requests used when interacting with REST APIs. While previous projects mainly focused on GET requests to retrieve data, today I learned how to create, update, and delete data using POST, PUT, and DELETE requests. To put these concepts into practice, I built a Habit Tracker using the Pixela API, where I created a user account, generated a graph, recorded my daily progress, automatically inserted the current date using Python's strftime() method, and learned how to edit or remove entries. This project provided hands-on experience with API authentication, HTTP headers, and CRUD (Create, Read, Update, Delete) operations—essential skills for working with modern web services.

```python
import requests 
PIXELA_ENDPOINT = "https://pixe.la/v1/users" 

user_params = { 
    "token": "", 
    "username": "", 
    "agreeTermsOfService": "yes", 
    "notMinor": "yes", 
    } 
response = requests.post( 
    url=PIXELA_ENDPOINT, 
    json=user_params 
    ) 
print(response.text)
```

Advanced Authentication Using HTTP Headers

HTTP headers are additional pieces of information sent along with an HTTP request. They provide metadata such as content type, authorization credentials, accepted formats, and more. Many modern REST APIs—including GitHub, Stripe, OpenAI, and Pixela—use headers to securely authenticate users without exposing sensitive information in the request body or URL.

```python
import requests 

USERNAME = "" 
TOKEN = "" 
GRAPH_ENDPOINT = f"https://pixe.la/v1/users/{USERNAME}/graphs"

headers = { 
    "X-USER-TOKEN": TOKEN 

} 

graph_config = { 
    "id": "graph1", 
    "name": "Coding Graph", 
    "unit": "hours", 
    "type": "float", 
    "color": "shibafu" 

} 

response = requests.post( 
    url=GRAPH_ENDPOINT, 
    json=graph_config, 
    headers=headers 

) 

print(response.text)
```

Add a Pixel to the Habit Tracker Using a POST Request

After creating our Pixela account and graph, the next step is to start tracking our daily progress. In Pixela, each daily record is represented by a pixel. Every time you complete your habit—for example, studying, exercising, reading, or coding—you add a new pixel to your graph.

```python
import requests 
from datetime import datetime 

USERNAME = "" 
TOKEN = "" 
GRAPH_ID = "" 

PIXEL_ENDPOINT = ( 
    f"https://pixe.la/v1/users/{USERNAME}/graphs/{GRAPH_ID}" 
) 

headers = { 
    "X-USER-TOKEN": TOKEN 

} 

today = datetime.now() 

pixel_data = { 
    "date": today.strftime("%Y%m%d"), 
    "quantity": "5" 

} 

response = requests.post(
     url=PIXEL_ENDPOINT, 
     json=pixel_data, 
     headers=headers 

) 

print(response.text)
```

Autofilling Today's Date Using strftime()

```python
import requests 
from datetime import datetime 

USERNAME = "" 
TOKEN = "" 
GRAPH_ID = "graph1" 

PIXEL_ENDPOINT = ( 
    f"https://pixe.la/v1/users/{USERNAME}/graphs/{GRAPH_ID}"
) 

headers = { 
    "X-USER-TOKEN": TOKEN 
} 

pixel_data = {
     "date": datetime.now().strftime("%Y%m%d"), 
     "quantity": "5" 

} 

response = requests.post( 
    url=PIXEL_ENDPOINT, 
    json=pixel_data, 
    headers=headers 
) 

print(response.text)
```

How to Use HTTP PUT and DELETE Requests

In the previous lessons, ie learned how to create a Pixela account, create a graph, and add daily habit records using HTTP POST requests. However, real-world applications often require modifying or removing existing data. This is where the PUT and DELETE HTTP methods become essential.

example:

```python
import requests 
from datetime import datetime 

USERNAME = "yourusername" 
TOKEN = "YOUR_TOKEN" 
GRAPH_ID = "graph1" 

headers = { 
    "X-USER-TOKEN": TOKEN 

} 

date = datetime.now().strftime("%Y%m%d") 

PIXEL_ENDPOINT = ( 
    f"https://pixe.la/v1/users/{USERNAME}/graphs/{GRAPH_ID}/{date}"


) 
response = requests.delete( 
    url=PIXEL_ENDPOINT, headers=headers 

)


print(response.text)
```
