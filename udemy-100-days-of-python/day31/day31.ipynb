## Creating the User Interface

```python
from tkinter import * 
BACKGROUND_COLOR = "#B1DDC6" 

window = Tk() 
window.title("Flashy") 
window.config(bg=BACKGROUND_COLOR, padx=50, pady=50) 
canvas = Canvas(width=800, height=526, bg=BACKGROUND_COLOR, highlightthickness=0) 
canvas.grid(row=0, column=0, columnspan=2) 
window.mainloop()
```

The canvas acts as the main display area where both the card images and vocabulary words appear.

### Loading Flash Cards
The application loads French vocabulary from a CSV file using Pandas.

```python
import pandas as pd 

data = pd.read_csv("french_words.csv") 
to_learn = data.to_dict(orient="records")
```

### Displaying a New Flash Card
Each time the user clicks the "Next" button, a random word is selected.


```python
import random

current_card = random.choice(to_learn) 

canvas.itemconfig(card_title, text="French") 
canvas.itemconfig(card_word, text=current_card["French"])
```

### Automatically Flipping the Card
After three seconds, the flashcard flips to reveal the English translation.


```python
def flip_card(): 
    canvas.itemconfig(card_title, text="English") 
    canvas.itemconfig(card_word, text=current_card["English"]) 
flip_timer = window.after(3000, flip_card)
```

### Preventing Multiple Timers
Whenever a new flashcard is displayed, the previous timer is canceled before creating another.


```python
window.after_cancel(flip_timer) 
flip_timer = window.after(3000, flip_card)
```

### Saving Learning Progress
When the user already knows a word, it is removed from the learning list.


```python

import pandas as pd 
data = pd.DataFrame(to_learn) 
data.to_csv("words_to_learn.csv", index=False)
```

### Loading Saved Progress

If a saved progress file exists, the application loads it. Otherwise, it starts with the original dataset.

```python
import pandas as pd 
try: 
    data = pd.read_csv("words_to_learn.csv") 
except FileNotFoundError: 
    data = pd.read_csv("french_words.csv") 
to_learn = data.to_dict(orient="records")
```
